```python

import json
from pathlib import Path

PROJECT_ROOT = Path(
    r"C:\Users\iso99\Desktop\Ahmed\current\Image-Captioning-RAG"
)

PROCESSED_DATA = PROJECT_ROOT / "02_processed_data"

expected = {
    "training": {"images": 9900, "captions": 49500},
    "validation": {"images": 100, "captions": 500},
    "test": {"images": 100, "captions": 500},
}

for split, counts in expected.items():

    split_dir = PROCESSED_DATA / split
    images_dir = split_dir / "images"
    annotations_dir = split_dir / "annotations"

    with open(
        annotations_dir / "captions.json",
        "r",
        encoding="utf-8"
    ) as f:
        captions = json.load(f)

    with open(
        annotations_dir / "instances.json",
        "r",
        encoding="utf-8"
    ) as f:
        instances = json.load(f)

    image_count = len(list(images_dir.glob("*.jpg")))
    caption_count = len(captions["annotations"])

    assert image_count == counts["images"], (
        f"{split}: incorrect image count"
    )

    assert len(captions["images"]) == counts["images"]
    assert caption_count == counts["captions"]

    assert len(instances["images"]) == counts["images"]

    print(
        f"{split}: PASS | "
        f"{image_count} images | "
        f"{caption_count} captions | "
        f"{len(instances['annotations'])} object annotations"
    )

print("\nAll extraction verification checks passed!")
```
